**Create a list that contains only even numbers from a range of numbers**

consider range of numbers between 560009 to 678456

    even numbers : 2,4,6,8,10
    odd numbers : 1,3,5,7,9,11

even numbers are all divisible by 2 and the remainder will be 0

```python
# Define an empty new list
even_nums = []
# Intialize i to 560009
i = 560009
# condition in while
while i<=678456:
    if i%2==0:
        even_nums.append(i)
    i+=1
```

```python
print(even_nums)
```
